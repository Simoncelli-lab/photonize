```python
import time

import event

# ---------------------------------------------------------------------------
# Input data
# Set "my_folder" to the directory containing your files; each of the three
# inputs is given by its full filename below. The Picasso .yaml sidecar of the
# localization file must sit next to it (the output .yaml is derived from it).
# ---------------------------------------------------------------------------
my_folder = r'/path/to/data'

# The three inputs evelyze expects:
#   1. picked (grouped) localizations (Picasso .hdf5, must contain a 'group' column)
#   2. photon index file (x, y, ms, dt; from the acquisition preprocessing)
#   3. drift file (.txt file, space-delimited x y per frame)
locs_file  = my_folder + '/' + r'locs_picked.hdf5'
index_file = my_folder + '/' + r'photon_index.hdf5'
drift_file = my_folder + '/' + r'drift.txt'

start_time = time.time()
event.evelyze(
    locs_file,
    index_file,
    drift_file,
    offset=10,                   # temporal oversampling factor of the localization file
    int_time=200,                # camera integration time per frame (ms)
    pos_diameter=4.5,            # ROI diameter (px) for position fitting
    lt_diameter=5.5,             # ROI diameter (px) for lifetime estimation
    frame_size=256,              # detector frame size (px); sets binning = 4096/frame_size
    ref_pixel_nm=115,            # binned pixel size (nm) = 7.2 x 4096/frame_size  (115 for 256 px)
    suffix='',
    max_dark_frames=1,           # max skipped frames when linking locs into an event
    proximity=2,                 # link threshold = proximity x (lpx + lpy)
    filter_single=True,          # drop single-localization events
    norm_brightness=True,        # normalize brightness across the field of view
    dt_window=None,              # optional (min, max) arrival-time gate
    more_ms=0,                   # extra ms cropped before/after each event
)
end_time = time.time()
print(f"Execution time: {end_time - start_time:.3f} seconds")
```
